## [Green City Script](https://custom-scripts.sentinel-hub.com/sentinel-2/green_city/) 

```python
evalscript = """
/*
Author of the script: Carlos Bentes
*/

// Normalized Difference Vegetation Index
var ndvi = (B08-B04)/(B08+B04);

// Threshold for vegetation
var veg_th = 0.4;

// Simple RGB
var R = 2.5*B04;
var G = 2.5*B03;
var B = 2.5*B02;

// Transform to Black and White
var Y = 0.2*R + 0.7*G + 0.1*B;
var pixel = [Y, Y, Y];

// Change vegetation color
if(ndvi >= veg_th)
  pixel = [0.1*Y, 1.8*Y, 0.1*Y];

return pixel;
"""
bbox = BBox(bbox=[-4.393185527198364, 55.78127484592951, -4.071705352343756, 55.92963532489984], crs=CRS.WGS84)

request = SentinelHubRequest(
    evalscript=evalscript,
    input_data=[
        SentinelHubRequest.input_data(
            data_collection=DataCollection.SENTINEL2_L2A,          
            time_interval=('2022-11-07', '2023-02-20'),          
            other_args={"dataFilter": {"maxCloudCoverage": 5,"mosaickingOrder": "leastCC"}}
        ),
    ],
    responses=[
        SentinelHubRequest.output_response('default', MimeType.TIFF),
    ],
    bbox=bbox,
    size=[512, 420.17],
    config=config
)

response = request.get_data()
image = response[0]
plot_image(image, factor=1/255, clip_range=(0,1))
```

## Downloading as Tiff

```python
print(ndvi_tiff.shape)
```

```python
import rasterio
from rasterio.transform import from_bounds

# Define bounding box (same as SentinelHub request)
minx, miny, maxx, maxy = -4.393185527198364, 55.78127484592951, -4.071705352343756, 55.92963532489984

# Extract TIFF data
ndvi_tiff = response[0]  # Your NDVI data
print(ndvi_tiff.shape)  # Check shape, should be (420, 512, 3)

# Unpack the shape to get height, width, and bands (3 channels)
height, width, bands = ndvi_tiff.shape

# Create the affine transform (maps pixels to coordinates)
transform = from_bounds(minx, miny, maxx, maxy, width, height)

# Define output file path
output_file = "./Glasgow/env_data/Greencity/gc_win_2023.tiff"

# Save the RGB image as GeoTIFF
with rasterio.open(output_file, "w",
                   driver="GTiff",
                   height=height,
                   width=width,
                   count=bands,  # 3 bands for RGB
                   dtype=ndvi_tiff.dtype,
                   crs="EPSG:4326",  # Assign correct CRS
                   transform=transform) as dst:  
    # Write each band separately (RGB channels)
    for i in range(bands):
        dst.write(ndvi_tiff[:, :, i], i + 1)

print(f"✅ Georeferenced NDVI GeoTIFF saved as {output_file}")
```

# Edinburgh Trees with Variable Diameter

```python
import geopandas as gpd
editrees = gpd.read_file("/Users/Isaiah/Documents/Dissertation/NoiseModelling/Edinburgh/env_data/Trees/Trees.shp")
```

```python
editrees.columns
```

```python
import geopandas as gpd
import pandas as pd
from shapely.geometry import Point

height_mapping = {
    "5 to 10 meters": 7.5,
    "10 to 15 meters": 12.5,
    "15 to 20 meters": 17.5,
    "20 to 25 meters": 22.5,
    "25 to 30 meters": 27.5,
    "Up to 5 meters": 4.5,
    "30 meters +": 32,
}

# Function to map height values
def extract_height(value):
    if pd.isna(value) or value is None:
        return None  # Keep NaN values

    value = str(value).strip()  # Convert to string and remove extra spaces
    
    if value in height_mapping:
        return height_mapping[value]  # Use predefined mapping
    
    return None  # If value is not found in mapping, return NaN

# Function for Diameter 
def extract_diameter(value):
    if pd.isna(value):
        return 0.15  # Default small tree diameter in meters if missing # there are issues here with the assumptions 

    value = str(value)  # Convert to string to avoid TypeError

    if " - " in value:
        try:
            low, high = map(float, value.split(" - "))
            return ((low + high) / 2) / 100  # Convert cm to meters
        except ValueError:
            return 0.15  # Default diameter for unexpected values
    
    elif value == "90 +":
        return 0.95  # Approximate for very large trees (converted to meters)
    
    elif value in ["Juvenile", "Semi-mature", "Middle Aged"]:
        return 0.10  # Assign small diameter for young trees (converted to meters)
    
    return 0.15  # Default diameter in meters if no match


# Apply functions to extract values
editrees["HEIGHT"] = editrees["Height"].apply(extract_height)  # Tree height (meters)

# Extract Diameter in meters
editrees["D_m"] = editrees["DiameterAt"].apply(extract_diameter)

# Create buffers using radius = D_m / 2
editrees["geometry"] = editrees["geometry"].buffer(editrees["D_m"] / 2) 


editrees["G"] = .9 # Wall absorption 
```

```python
editrees.head()
```

```python
# Assuming your GeoDataFrame is named 'gdf'
editrees['HEIGHT'] = editrees['HEIGHT'].fillna(5)
```

```python
# Define the columns I want to keep
keep_col = ["OBJECTID", "HEIGHT", "geometry", "G"]
# Keep only the relevant columns
editrees = editrees[keep_col]
```
