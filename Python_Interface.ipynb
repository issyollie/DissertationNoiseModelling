### Importing ASC FOLDER 
Your error message suggests that there is an issue with reading the ASC files, specifically with parsing floating-point values like 78.8339996337890625. Here are a few potential reasons and solutions:

Possible Causes & Fixes:
1. Corrupted or Improperly Formatted ASC Files
The ASC files might have unexpected characters, incorrect delimiters, or invalid data entries.
Fix: Open the ASC files in a text editor or GIS software (like QGIS) to inspect the format and ensure they follow the ESRI ASCII Raster format correctly.
2. Decimal Separator Issue
Some systems use commas (78,8339996) instead of dots (78.8339996) for decimal separators.
Fix: Ensure your ASC files use a dot (.) as the decimal separator.
3. Projection File (.prj) Issues
The script tries to read the .prj file to get the projection information. If the .prj file is missing or corrupted, it might cause issues.
Fix: Ensure that each ASC file has a corresponding .prj file with valid content.
4. Database Writing Error
The error message references java.sql.SQLException, which could mean the database rejected the input due to unexpected characters.
Fix: Try manually importing a single ASC file using a database query to check if the issue persists.
5. Large File Handling
If the ASC files are large, there could be memory issues.
Fix: Try downscaling (downscale > 1) to see if smaller file sizes work.
Next Steps:
Check an ASC file manually—open it and inspect its values.
Ensure correct decimal separator—values should use dots (.), not commas.
Try importing a single ASC file instead of a folder.
Enable debugging logs in the Groovy script to capture more detailed errors.
Try with a sample file to isolate the issue.

## Average Humidity July 2022 

```python
import requests
import pandas as pd
from io import StringIO
# Define the API URL
url = "https://power.larc.nasa.gov/api/temporal/hourly/point"
params = {
    'start': '20220101', # Use this for the beginning date 
    'end': '20221231', # use this for the end date 
    'latitude': '55.953251', # 
    'longitude': '-3.188267',    'community': 're', # re means renewable energy 
    'parameters': 'RH2M', # code for wind relative humidity at 2 meters 
    'format': 'csv', 
    'user': 'issyollie',
    'header': 'false',
    'time-standard': 'utc'
}
# Send GET request to the API
response = requests.get(url, params=params)

# Check if the request was successful (status code 200)
if response.status_code == 200:
    # Read the CSV data from the response into a pandas DataFrame
    data = StringIO(response.text)  # Convert string response to a file-like object
    RH2M_df = pd.read_csv(data) # read as a dataframe 
    
    # Display the first few rows of the DataFrame
    print(RH2M_df.head())
else:
    print(f"Failed to fetch data. Status code: {response.status_code}")

import pandas as pd
# Assuming the dataframe is named 'df'
# Filter the rows where the month is 7
july_data = RH2M_df[RH2M_df['MO'] == 7]
dec_data = RH2M_df[RH2M_df['MO'] == 12]
# Calculate the average temperature for July
july_2022_RH2M = july_data['RH2M'].mean()
dec_2022_RH2M = dec_data['RH2M'].mean()
```

```python
print (dec_2022_RH2M, july_2022_RH2M)
```
